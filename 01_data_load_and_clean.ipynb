**Column name Corrected data set**

```python
import pandas as pd

col_names = [
    "SMI",
    "lambda1_sTDA_nm",
    "F1_sTDA",
    "lambda1_TDDFT_nm",
    "F1_TDDFT",
    "lambda_max_exp_nm",
    "extinction",
    "solvent"
]

df = pd.read_csv("paper_allDB.csv", header=None, skiprows=1, names=col_names)

print(df.shape)
print(df.dtypes)
print(df.isna().sum())
print(df.head())
```

**Plotting Lambda max**

```python
import matplotlib.pyplot as plt

# Filter to rows with a valid experimental lambda_max
df_exp = df[df["lambda_max_exp_nm"].notna()].copy()

print(f"Rows with experimental lambda_max: {df_exp.shape[0]}")
print(df_exp["lambda_max_exp_nm"].describe())

plt.figure(figsize=(8, 5))
plt.hist(df_exp["lambda_max_exp_nm"], bins=100, edgecolor="black")
plt.xlabel("Experimental λmax (nm)")
plt.ylabel("Count")
plt.title("Distribution of experimental λmax across dataset")
plt.show()
```

**restrict to a physically sane range for standard UV-Vis spectroscopy**

```python
# Physically reasonable range for standard UV-Vis-NIR spectroscopy
df_clean = df_exp[(df_exp["lambda_max_exp_nm"] >= 200) & (df_exp["lambda_max_exp_nm"] <= 900)].copy()

print(f"Rows before range filter: {df_exp.shape[0]}")
print(f"Rows after range filter: {df_clean.shape[0]}")
print(f"Rows dropped: {df_exp.shape[0] - df_clean.shape[0]}")

plt.figure(figsize=(8, 5))
plt.hist(df_clean["lambda_max_exp_nm"], bins=100, edgecolor="black")
plt.xlabel("Experimental λmax (nm)")
plt.ylabel("Count")
plt.title("Distribution of experimental λmax (filtered, 200-900 nm)")
plt.show()
```

**look at what's actually piling up at those spikes: 250, 345**

```python
# Find the exact bin edges responsible for those two spikes
print(df_clean["lambda_max_exp_nm"].value_counts().head(15))
```

**Checking cluster at 254 nm**

```python
# Look at the SMILES diversity within the suspicious 254 nm cluster
cluster_254 = df_clean[df_clean["lambda_max_exp_nm"] == 254.0]
print(f"Number of rows at exactly 254.0 nm: {cluster_254.shape[0]}")
print(f"Number of UNIQUE SMILES among them: {cluster_254['SMI'].nunique()}")
print(cluster_254["SMI"].head(10).to_list())
```

```python
# Drop the suspicious 254.0 nm cluster (likely HPLC detection wavelength artifact)
df_clean2 = df_clean[df_clean["lambda_max_exp_nm"] != 254.0].copy()

print(f"Rows before dropping 254nm cluster: {df_clean.shape[0]}")
print(f"Rows after: {df_clean2.shape[0]}")
```

**Rechecking**

```python
from rdkit import Chem

df_clean2["mol"] = df_clean2["SMI"].apply(Chem.MolFromSmiles)
df_clean2["is_valid"] = df_clean2["mol"].notna()

print(f"Total rows: {df_clean2.shape[0]}")
print(f"Valid SMILES: {df_clean2['is_valid'].sum()}")
print(f"Invalid SMILES: {(~df_clean2['is_valid']).sum()}")

invalid_examples = df_clean2[~df_clean2["is_valid"]]["SMI"].head(5)
print(invalid_examples)
```

**Dropping anomalus rows**

```python
df_valid = df_clean2[df_clean2["is_valid"]].copy()
print(f"Rows before SMILES validation: {df_clean2.shape[0]}")
print(f"Rows after dropping invalid SMILES: {df_valid.shape[0]}")
```

**check for duplicate compounds**

```python
df_valid["canonical_smi"] = df_valid["mol"].apply(Chem.MolToSmiles)

n_total = df_valid.shape[0]
n_unique = df_valid["canonical_smi"].nunique()
print(f"Total valid rows: {n_total}")
print(f"Unique canonical structures: {n_unique}")
print(f"Duplicate rows: {n_total - n_unique}")
```

**Check agreement**

```python
# Look at how much duplicate reports of the same compound actually disagree
dup_mask = df_valid.duplicated(subset="canonical_smi", keep=False)
dup_groups = df_valid[dup_mask].groupby("canonical_smi")["lambda_max_exp_nm"].agg(["count", "mean", "std", "min", "max"])
dup_groups_sorted = dup_groups.sort_values("std", ascending=False)

print(dup_groups_sorted.head(10))
```

**Flag duplicate groups by how much they disagree**

```python
# Flag duplicate groups by how much they disagree
dup_groups_sorted["range"] = dup_groups_sorted["max"] - dup_groups_sorted["min"]

# A defensible physical cutoff: >100 nm disagreement is not plausible solvatochromism
suspicious = dup_groups_sorted[dup_groups_sorted["range"] > 100]
trustworthy = dup_groups_sorted[dup_groups_sorted["range"] <= 100]

print(f"Duplicate groups total: {dup_groups_sorted.shape[0]}")
print(f"Suspicious (>100 nm spread): {suspicious.shape[0]}")
print(f"Trustworthy (<=100 nm spread): {trustworthy.shape[0]}")
```

**7,167 unique compounds with trustworthy experimental λmax values.**

```python
# Get the canonical_smi values in each category
suspicious_smis = suspicious.index.tolist()
trustworthy_smis = trustworthy.index.tolist()

# Rule 1: drop all rows belonging to suspicious duplicate groups entirely
df_no_suspicious = df_valid[~df_valid["canonical_smi"].isin(suspicious_smis)].copy()

# Rule 2: collapse trustworthy duplicates by averaging lambda_max, keep one row per compound
# First, separate non-duplicated rows (they pass through untouched)
non_dup_mask = ~df_no_suspicious.duplicated(subset="canonical_smi", keep=False)
df_singles = df_no_suspicious[non_dup_mask].copy()

# Then collapse the trustworthy duplicate groups
df_trustworthy_dups = df_no_suspicious[df_no_suspicious["canonical_smi"].isin(trustworthy_smis)]
df_collapsed = (
    df_trustworthy_dups
    .groupby("canonical_smi", as_index=False)
    .agg({
        "SMI": "first",              # keep one representative raw SMILES
        "lambda_max_exp_nm": "mean", # average the duplicate measurements
        "extinction": "mean",
        "solvent": "first"
    })
)

# Combine singles + collapsed duplicates into final Phase 1 dataset
df_final = pd.concat([df_singles, df_collapsed], ignore_index=True, sort=False)

print(f"Starting valid rows: {df_valid.shape[0]}")
print(f"After dropping suspicious duplicates: {df_no_suspicious.shape[0]}")
print(f"Final rows (singles + collapsed duplicates): {df_final.shape[0]}")
print(f"Unique compounds check: {df_final['canonical_smi'].nunique() if 'canonical_smi' in df_final.columns else 'N/A - check columns'}")
```

**Final check**

```python
print(df_final.columns.tolist())
print(df_final.isna().sum())
print(df_final["lambda_max_exp_nm"].describe())
```

```python
# Keep only the columns we actually need going forward
df_export = df_final[["SMI", "canonical_smi", "lambda_max_exp_nm", "extinction", "solvent"]].copy()
df_export.to_csv("beard_uvvis_cleaned.csv", index=False)
print(f"Saved {df_export.shape[0]} rows to beard_uvvis_cleaned.csv")
```
